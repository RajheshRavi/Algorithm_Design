## Why do we need a Function ?

Functions let us group a task into a **named, reusable block of code**. Instead of repeating the same instructions throughout a program, we write them once in a function and call it whenever needed.

They make programs easier to **read, test, debug, and maintain**. Functions also support *modularity*: different parts of a program can focus on specific responsibilities, while passing inputs and returning results when necessary.

### Example Problem Statement:

Consider the situation where there are 5 arrays, and we need to search element-1 in array-1, element-2 in array-2 and so on.

The Plain solution would look like:

```python
a1 = [i for i in range(1, 10)]
a2 = [i for i in range(11, 40) if i % 2 == 0]
a3 = [i for i in range(31, 100) if i % 3 == 0]
a4 = [i for i in range(1, 100) if i % 5 == 0]
a5 = [i for i in range(1, 150) if i % 7 == 0]
```

```python
# Search for the number 8 in the list a1 and print it if found
for i in a1:
    if 8 == i:
        print(i)

# Search for the number 20 in the list a2 and print it if found
for i in a2:
    if 20 == i:
        print(i)

# Search for the number 54 in the list a3 and print it if found
for i in a3:
    if 54 == i:
        print(i)

# Search for the number 50 in the list a4 and print it if found
for i in a4:
    if 50 == i:
        print(i)

# Search for the number 140 in the list a5 and print it if found
for i in a5:
    if 140 == i:
        print(i)
    
```

### Do you notice anything common in the above Cell?

This is where Function comes in hand to help us reduce code duplication.

```python
def function():
    print("Hello World")
```

```python
function()
```

### What can we do if we need some inference of the data or state even after the function executed has completed?

Return statement helps us overcome this issue

```python
def find_element(element_list, element):
    for i in element_list:
        if element == i:
            return True
    return False
```

```python
print(find_element(a1, 8))  # Should return True
print(find_element(a2, 20))  # Should return True
print(find_element(a3, 54))  # Should return True
print(find_element(a4, 50))  # Should return True
print(find_element(a5, 140))  # Should return True
```

What if we search for an element not present in the list?

```python
print(find_element(a1, 11))  # Should return False
```

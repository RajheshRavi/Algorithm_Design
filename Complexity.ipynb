## Introduction to Algorithmic Complexity
### What is Algorithmic Complexity?
Algorithmic Complexity describes how the resources required by an algorithm grow as the size of its input grows. The two resources considered most often are execution time and memory.

Rather than measuring exact seconds or exact bytes, complexity focuses on the rate of growth. This makes it possible to compare algorithms independently of a particular programming language, machine, or hardware configuration.

If the input size is represented by _"n"_, algorithmic complexity expresses how the time or memory requirements change as _"n"_ increases.

### Why do we need Algorithmic Complexity?
We use algorithmic complexity to assess whether an application can perform reliably and efficiently at its expected scale.

It helps us estimate resource needs during design, development, testing, and production planning. For example, an algorithm that performs well for 100 records may become too slow or consume too much memory when processing millions of records.

Complexity analysis helps teams make informed decisions about:

- Performance: ensuring operations complete within acceptable response-time limits.
- Scalability: predicting how an application behaves when data volume or user traffic increases.
- Capacity planning: provisioning appropriate compute and memory resources.
- Design choices: selecting suitable data structures and algorithms.
- Cost optimisation: avoiding unnecessary cloud infrastructure or inefficient processing.

### How do we calculate Algorithmic Complexity?
Algorithmic complexity is calculated by analysing an algorithm’s control flow and determining how many operations it performs, or how much additional memory it requires, relative to the input size 
_"n"_.

The analysis usually considers the following:

- The number of loop iterations.
- Whether loops are sequential or nested.
- Whether the input is repeatedly divided into smaller parts.
- The amount of additional memory created while processing the input.
- The best-case, average-case, or worst-case behaviour of the algorithm.
- For most high-level design and performance discussions, we are primarily interested in the worst-case growth rate, because it provides an upper bound on the resources the algorithm may require.

### Asymptotic Notations

- $ O ( n ) $ - Big O
- $ \Omega ( n ) $ - Big Theta
- $ \Theta ( n ) $ - Big Theta
- $ o ( n ) $ - Small O
- $ \omega ( n ) $ - - Small Omega


## Difference Between $O$, $\Omega$, $\Theta$, $o$ and $\omega$ 


| Notation | Meaning | Bound Type |
|---|---|---|
| $$O(g(n))$$ | Grows **at most as fast as** $$g(n)$$ | **Upper bound** |
| $$\Omega(g(n))$$ | Grows **at least as fast as** $$g(n)$$ | **Lower bound** |
| $$\Theta(g(n))$$ | Grows **at the same rate as** $$g(n)$$ | **Tight bound** |
| $$o(g(n))$$ | Grows **strictly slower than** $$g(n)$$ | **Strict upper bound** |
| $$\omega(g(n))$$ | Grows **strictly faster than** $$g(n)$$ | **Strict lower bound** |



## Common Complexity Classes

| Complexity | Name | Typical Example |
|---|---|---|
| $$O(1)$$ | **Constant** | Accessing an array element by index |
| $$O(\log n)$$ | **Logarithmic** | Binary search on sorted data |
| $$O(n)$$ | **Linear** | Scanning each item in a list once |
| $$O(n \log n)$$ | **Linearithmic** | Efficient sorting algorithms, such as merge sort |
| $$O(n^2)$$ | **Quadratic** | Comparing every item with every other item |
| $$O(2^n)$$ | **Exponential** | Brute-force evaluation of all combinations |
| $$O(n!)$$ | **Factorial** | Evaluating all possible permutations |

In general, lower complexity growth rates are more scalable for large inputs:

```math
O(1) < O(\log n) < O(n) < O(n \log n) < O(n^2) < O(2^n) < O(n!)

# Asymptotic Notations of Complexity

## 1. Introduction

Asymptotic notations are mathematical tools used to analyze the time and space complexity of algorithms as the input size increases.

Let $n$ represent the input size and $T(n)$ represent the running time of an algorithm.

The four important asymptotic notations are:

- **Big O:** $O(g(n))$ — Asymptotic upper bound.
- **Small o:** $o(g(n))$ — Strict asymptotic upper bound.
- **Big Omega:** $\Omega(g(n))$ — Asymptotic lower bound.
- **Big Theta:** $\Theta(g(n))$ — Tight asymptotic bound.

## 2. Big O Notation — $O$

Big O notation describes an asymptotic upper bound on the growth of a function.

### Mathematical Definition

A function $T(n)$ belongs to $O(g(n))$ if there exist positive constants $c$ and $n_0$ such that

$$
0 \leq T(n) \leq c g(n),
\quad \forall n \geq n_0
$$

### Example

Consider the function

$$
T(n) = 3n^2 + 5n + 2
$$

For $n \geq 1$,

$$
3n^2 + 5n + 2 \leq 10n^2
$$

Therefore,

$$
T(n) = O(n^2)
$$

**Interpretation:** The function grows no faster than a constant multiple of $n^2$ for sufficiently large $n$.

## 3. Small o Notation — $o$

Small o notation describes a strict asymptotic upper bound. It indicates that a function grows strictly slower than another function.

### Mathematical Definition

A function $T(n)$ belongs to $o(g(n))$ if, for every constant $c > 0$, there exists $n_0$ such that

$$
0 \leq T(n) < c g(n),
\quad \forall n \geq n_0
$$

An equivalent limit definition is

$$
\lim_{n \to \infty} \frac{T(n)}{g(n)} = 0
$$

### Example

Let

$$
T(n) = 5n + 10
$$

Compare it with $g(n) = n^2$:

$$
\lim_{n \to \infty} \frac{5n+10}{n^2}
=
\lim_{n \to \infty}
\left(\frac{5}{n}+\frac{10}{n^2}\right)
= 0
$$

Therefore,

$$
5n + 10 = o(n^2)
$$

However,

$$
n^2 \notin o(n^2)
$$

because the ratio $n^2/n^2$ approaches $1$, not $0$.

**Interpretation:** Small o represents a strictly slower rate of growth.

## 4. Big Omega Notation — $\Omega$

Big Omega notation describes an asymptotic lower bound on the growth of a function.

### Mathematical Definition

A function $T(n)$ belongs to $\Omega(g(n))$ if there exist positive constants $c$ and $n_0$ such that

$$
0 \leq c g(n) \leq T(n),
\quad \forall n \geq n_0
$$

### Example

Consider

$$
T(n) = 3n^2 + 5n + 2
$$

For $n \geq 1$,

$$
T(n) \geq 3n^2
$$

Therefore,

$$
T(n) = \Omega(n^2)
$$

**Interpretation:** The function grows at least as fast as a positive constant multiple of $n^2$ for sufficiently large $n$.

## 5. Big Theta Notation — $\Theta$

Big Theta notation describes a tight asymptotic bound. It means that a function has both an asymptotic upper bound and an asymptotic lower bound of the same order.

### Mathematical Definition

A function $T(n)$ belongs to $\Theta(g(n))$ if there exist positive constants $c_1$, $c_2$, and $n_0$ such that

$$
0 \leq c_1 g(n) \leq T(n) \leq c_2 g(n),
\quad \forall n \geq n_0
$$

Equivalently,

$$
T(n) \in \Theta(g(n))
\iff
T(n) \in O(g(n))
\text{ and }
T(n) \in \Omega(g(n))
$$

### Example

Consider

$$
T(n) = 3n^2 + 5n + 2
$$

For $n \geq 1$,

$$
3n^2 \leq T(n) \leq 10n^2
$$

Thus, choosing $c_1 = 3$ and $c_2 = 10$, we obtain

$$
T(n) = \Theta(n^2)
$$

**Interpretation:** The function grows at the same asymptotic rate as $n^2$, up to constant factors.

## 6. Comparison of Asymptotic Notations

| Notation | Meaning | Example |
|---|---|---|
| $O(g(n))$ | Upper bound | $n = O(n^2)$ |
| $o(g(n))$ | Strict upper bound | $n = o(n^2)$ |
| $\Omega(g(n))$ | Lower bound | $n^2 = \Omega(n)$ |
| $\Theta(g(n))$ | Tight bound | $3n^2+5n+2 = \Theta(n^2)$ |

## 7. Common Orders of Growth

The following are common complexity classes, arranged approximately from slower to faster growth:

| Complexity | Name | Example |
|---|---|---|
| $O(1)$ | Constant | Array element access |
| $O(\log n)$ | Logarithmic | Binary search |
| $O(n)$ | Linear | Array traversal |
| $O(n\log n)$ | Linearithmic | Merge sort |
| $O(n^2)$ | Quadratic | Two nested loops |
| $O(2^n)$ | Exponential | Enumerating all subsets |

## 8. Important Observations

1. **Constant factors are ignored:**

   $$
   7n = \Theta(n)
   $$

2. **Lower-order terms are ignored when finding the tight bound:**

   $$
   n^2 + 4n + 10 = \Theta(n^2)
   $$

3. **Big O is not necessarily the tightest bound:**

   $$
   n = O(n^2)
   $$

   but the tighter bound is $n = \Theta(n)$.

4. **Small o is stricter than Big O:**

   $$
   n = o(n^2), \qquad n \notin o(n)
   $$

5. **Big Omega represents a lower bound**, not necessarily the best-case running time.

6. **Big Theta represents a tight bound** when both upper and lower bounds have the same growth rate.

## 9. Conclusion

Asymptotic notations help us understand how algorithmic resource requirements grow with input size.

- $O(g(n))$ describes an upper bound.
- $o(g(n))$ describes a strict upper bound.
- $\Omega(g(n))$ describes a lower bound.
- $\Theta(g(n))$ describes a tight bound.

These notations are fundamental tools for comparing algorithms and evaluating their efficiency and scalability.
